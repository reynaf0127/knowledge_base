11. Your task is to compute the investment performance of the Walbright Fund for the most recent year. The facts are as follows:  
■ On 1 January, the Walbright Fund had a market value of USD100 million.  
■ During the period 1 January to 30 April, the stocks in the fund generated a capital gain of USD10 million.  
■ On 1 May, the stocks in the fund paid a total dividend of USD2 million. All dividends were reinvested in additional shares.  
■ Because the fund’s performance had been exceptional, institutions invested an additional USD20 million in Walbright on 1 May, raising assets under management to USD132 million (USD100 + USD10 + USD2 + USD20).  
■ On 31 December, Walbright received total dividends of USD2.64 million. The fund’s market value on 31 December, not including the USD2.64 million in dividends, was USD140 million.  
■ The fund made no other interim cash payments during the year.

```python
# Calculate time-weighted return
amount_invested = np.array([100,112+20])
ending_value = np.array([112,140+2.64])
returns = (ending_value - amount_invested) / amount_invested
twrr = np.prod(1 + returns) - 1
print(returns)
print("TWRR:", twrr)
```

```python
# Calcualte money-weighted return
CF0 = -100
CF1 = -20
CF2 = 0
CF3 = 142.64
def npv(r):
    return CF0 + CF1/(1+r) + CF2/(1+r)**2 + CF3/(1+r)**3
irr_guess = 0.2
irr = fsolve(npv, irr_guess)
# annualize the IRR
irr_annual = (1+irr[0]) ** 3 - 1
print("IRR:", irr[0])
print("Annualized IRR:", irr_annual)
```

difference between time-weighted return (21.03%) vs money-weighted return (20.05%). the fund performance was relatively poorer during the eight month period, when the fund had more money invested, than the overall performance. this fact is reflected in a lower money-weighted rate of return compared with the time-weighted rate of return, as the money-weighted return is sensitive to the timeing and amount of withdarawals and additions to the portfolio

3. **Periodic interest**: a 20 year bond was issued at an annualized coupon rate of 6.7% and YTM of 6.7% and the coupon payments are semiannual

```python
# the PV of the bond at issuance
FV = 100
r = 0.067
t = 20
PMT = 100 * 0.067 / 2 # semi-annual payment
def pv_annuity(PMT, r, t):
    pv = 0
    for i in range(1, t):
        pv += PMT / (1 + r/2)**i # semi-annual payment
    return pv
# Example
PV = pv_annuity(PMT=PMT, r=r, t=t*2)
pv_final = PV + (PMT + FV)/((1+r/2)**(t*2))
print("Present Value:", pv_final)
```

9. **Coupon bond implied return**: seven-year Greek government bonds issued in 2020 with a 2% annual coupon had a price of EUR93.091 per EUR100 principal two years later.

```python
# What is the implied two-year return for an investor able to reinvest periodic interest at the original YTM of 2.00 percent?
ytm = 0.02
t = 2
PV = 100
FV = 93.091
def equation(r):
    PV_calculate = 0
    PMT = PV * ytm
    for i in range(1, t):
        PV_calculate += PMT/((1+r)**i)
    PV_final = PV_calculate + (PMT+FV)/((1+r)**t)
    return (PV_final - PV)
# initial guess
r_guess = 0.1
r_solution = fsolve(equation, r_guess)
print("r:", r_solution[0])
```

14. **Put option**: A put option grants the owner the right to sell a stock at a predetermined exercise price X. If we assume X = CNY50, then the put option value in one period (p1) is equal to Max(0, CNY50 − S1).

```python
# Using a one-period binomial tree model with the same prices in one year, initial stock price (S0) of CNY40, and 5 percent discount rate, create a risk-free portfolio replicating the put option with 0.75 units of the underlying asset and solve for the put option price (p0).
S0 = 40
X = 50
r=0.05 #discount rate
u=0.75
increase_r=0.4
decrease_r=0.2
increase_s1=S0*(1+increase_r)
decrease_s1=S0*(1-decrease_r)
#put option is not sell if price increase while call option is not buy if price decrease
p1_increase = max(0, X-increase_s1) 
p1_decrease = max(0, X-decrease_s1)
# vo = u*S0+p0
v1_increase = u*increase_s1+p1_increase
v1_decrease = u*decrease_s1+p1_decrease
print("v1_increase=v1_descrease")
v0 = v1_increase/(1+r)
# vo = u*S0+p0
p0 = v0 - u*S0
print("Put Option Price:", p0)
```

```python
# Grupo Ignacia issued 10-year corporate bonds two years ago. The bonds pay an annualized coupon of 10.7 percent on a semiannual basis, and the current annualized YTM is 11.6 percent. The current price of Grupo Ignacia’s bonds (per MXN100 of par value) is closest to:
n = 10
n_current = 2
n_left = n - n_current
coupon_rate = 0.107
ytm = 0.116
FV = 100
PMT = FV * coupon_rate / 2 # semi-annual payment
def pv_annuity(PMT, r, t):
    pv = 0
    for i in range(1, t*2):
        pv += PMT / (1 + r/2)**i # semi-annual payment
    return pv
PV = pv_annuity(PMT=PMT, r=ytm, t=n_left) + (PMT + FV)/((1+ytm/2)**(n_left*2))
print("Current Price of the Bond:", PV)
```

```python
# Grupo Ignacia issued 10-year corporate bonds four years ago. The bonds pay an annualized coupon of 10.7 percent on a semiannual basis, and the current price of the bonds is MXN97.50 per MXN100 of par value. The YTM of the bonds is closest to:
n = 10 - 4
r = 0.107
n_total = n * 2
pv = 97.5
FV = 100
PMT = FV * r / 2
def equation(ytm):
    pv_calculate = 0
    for i in range(1, n_total):
        pv_calculate += PMT / (1 + ytm/2)**i
    pv_final = pv_calculate + (PMT + FV) / (1 + ytm/2)**n_total
    return pv_final - pv
ytm_guess = 0.1
ytm_solution = fsolve(equation, ytm_guess)
print("YTM:", ytm_solution[0])
```
